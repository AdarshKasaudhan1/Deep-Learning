#Project 8 : Gradient Descent Implementation

1. Let's import the libraries

First, let's import the libraries we need for calculations, handling data, and plotting the results.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

2. Let's create a simple dataset

Here, we are using a small dataset where the output follows a simple linear pattern. We will use this data to train our model and see how well it learns.

```python
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

print("Input X:", X)
print("Actual y:", y)
```

3. Let's initialize the parameters

Before training, we set the weight and bias to zero. These are the parameters that our model will update during Gradient Descent.

```python
w = 0.0
b = 0.0

learning_rate = 0.01
iterations = 100

print("Initial Weight:", w)
print("Initial Bias:", b)
print("Learning Rate:", learning_rate)
```

4. Let's implement Gradient Descent

Now, let's implement Gradient Descent step by step. The model makes predictions, calculates the loss, finds the gradients, and updates the weight and bias in each iteration.

The main goal is to reduce the difference between the actual values and the predicted values.

```python
loss_history = []

n = len(X)

for i in range(iterations):

    # Forward pass
    y_pred = w * X + b

    # Calculate Mean Squared Error
    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    # Calculate gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Final Weight:", w)
print("Final Bias:", b)
print("Final Loss:", loss_history[-1])
```

5. Let's check the training process

Here, we store the loss values during training so we can see how the model improves over multiple iterations.

```python
results_df = pd.DataFrame({
    "Iteration": range(1, iterations + 1),
    "Loss": loss_history
})

results_df.head(100)
```

6. Let's visualize the loss reduction

The graph shows how the loss changes as the number of iterations increases. A decreasing loss means the model is learning better parameter values and making more accurate predictions.

```python
plt.figure(figsize=(8, 5))

plt.plot(
    results_df["Iteration"],
    results_df["Loss"]
)

plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Gradient Descent: Loss Reduction")
plt.grid(True)

plt.show()
```

```python
print("Final Model:")
print(f"y = {w:.2f}x + {b:.2f}")
```

7. Let's check the final parameters

After training, we display the final weight, bias, and loss. These values help us understand how well the model learned from the dataset.

```python
comparison_df = pd.DataFrame({
    "Parameter": [
        "Initial Weight",
        "Final Weight",
        "Initial Bias",
        "Final Bias",
        "Initial Loss",
        "Final Loss"
    ],
    "Value": [
        0,
        w,
        0,
        b,
        loss_history[0],
        loss_history[-1]
    ]
})

comparison_df
```

As training progressed, the loss decreased, showing that the model was improving its predictions. This helped me understand how Gradient Descent updates model parameters to reduce prediction error.

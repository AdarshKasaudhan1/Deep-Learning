Dataset: Pima Indians Diabetes Database

Rows: 768

Features: 8 numerical

Target: Outcome (0/1)

Why it's ideal: Works perfectly for testing hidden-layer activation functions (ReLU vs Sigmoid vs Tanh) while keeping the output layer as Sigmoid.

**Cell 1: Install & Import Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Cell 2: Load Dataset**

```python
df = pd.read_csv("/diabetes.csv")

print("Dataset Shape:", df.shape)
df.head()
```

**Cell 3: Prepare Features & Target**

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Features:", X.columns.tolist())
print("Target:", y.name)
```

**Cell 4: Define Activation Functions**

```python
# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# ReLU
def relu(x):
    return np.maximum(0, x)

# Tanh
def tanh(x):
    return np.tanh(x)
```

```python
import pandas as pd
import numpy as np

# Different input values
sample_input = np.array([-5, -2, -1, 0, 1, 2, 5])

comparison = pd.DataFrame({
    "Input": sample_input,
    "Sigmoid": sigmoid(sample_input),
    "ReLU": relu(sample_input),
    "Tanh": tanh(sample_input)
})

print(comparison)
```

**Cell 5: Compare Activation Functions Graphically**

```python
x = np.linspace(-10, 10, 500)

plt.figure(figsize=(8, 8))

plt.subplot(3,1,1)
plt.plot(x, sigmoid(x), color="blue")
plt.title("Sigmoid Activation Function")
plt.grid(True)

plt.subplot(3,1,2)
plt.plot(x, relu(x), color="green")
plt.title("ReLU Activation Function")
plt.grid(True)

plt.subplot(3,1,3)
plt.plot(x, tanh(x), color="red")
plt.title("Tanh Activation Function")
plt.grid(True)

plt.tight_layout()
plt.show()
```

**Cell 6: Apply Functions on Dataset**

```python
sigmoid_output = sigmoid(X)
relu_output = relu(X)
tanh_output = tanh(X)

print("Original Features")
display(X.head())

print("Sigmoid Output")
display(sigmoid_output.head())

print("ReLU Output")
display(relu_output.head())

print("Tanh Output")
display(tanh_output.head())
```

**Cell 8: Summary Statistics Comparison**

```python
comparison = pd.DataFrame({
    "Original Mean": X.mean(),
    "Sigmoid Mean": sigmoid_output.mean(),
    "ReLU Mean": relu_output.mean(),
    "Tanh Mean": tanh_output.mean()
})

comparison
```

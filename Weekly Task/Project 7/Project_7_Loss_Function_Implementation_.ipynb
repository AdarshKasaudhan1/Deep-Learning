#7. Loss Function Implementation

1. Let’s import the library

We’ll use NumPy to perform calculations and find out how much the predicted values differ from the actual values.

```python
import numpy as np
```

2. Let’s set some actual and predicted values

We’ll use a simple binary classification example. The actual values are either 0 or 1, while the predicted values represent the probability of belonging to class 1.

```python
# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_predicted = np.array([0.9, 0.2, 0.8, 0.6, 0.1])

print("Actual Values:   ", y_actual)
print("Predicted Values:", y_predicted)
```

3. Let’s calculate Mean Squared Error (MSE)

MSE calculates the average of the squared differences between the actual values and the predicted values. It helps us understand how far the predictions are from the correct answers.

```python
mse = np.mean((y_actual - y_predicted) ** 2)

print(f"Mean Squared Error: {mse:.4f}")
```

4. Let’s calculate Cross-Entropy Loss

Cross-entropy loss measures how well the predicted probabilities match the actual class labels. We’ll use the predicted probabilities to calculate the loss, while avoiding errors caused by taking the logarithm of zero.

```python
# Avoid log(0)
epsilon = 1e-15

y_predicted_clipped = np.clip(y_predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_actual * np.log(y_predicted_clipped) +
    (1 - y_actual) * np.log(1 - y_predicted_clipped)
)

print("Cross-Entropy Loss:", cross_entropy)
```

5. Let’s compare both loss functions

Now we’ll display the MSE and cross-entropy loss values together. This helps us see how the two methods measure prediction errors differently.

```python
print("Loss Function Comparison")
print("------------------------")
print(f"Mean Squared Error : {mse:.4f}")
print(f"Cross-Entropy Loss : {cross_entropy:.4f}")
```

6. What’s the difference between MSE and cross-entropy?

We’ll create a simple table showing the typical use and purpose of each loss function.

- MSE: Commonly used in regression problems to measure the difference between actual and predicted numerical values.

- Cross-Entropy Loss: Commonly used in classification problems to measure how well predicted probabilities match the correct classes.

In classification, cross-entropy is generally preferred because it directly evaluates the predicted probabilities and penalizes confident incorrect predictions.

```python
import pandas as pd

df = pd.DataFrame({
    "Loss Function": ["Mean Squared Error (MSE)", "Cross-Entropy Loss"],
    "Typical Use": ["Regression", "Classification"],
    "Main Purpose": ["Measures the squared difference between actual and predicted values",
        "Measures the difference between actual class labels and predicted probabilities"],

    "Role in Neural Networks": ["Helps minimize prediction errors in regression problems",
        "Helps optimize classification predictions"]
})

df
```
